```python
from transformers import pipeline
generator = pipeline("text-generation", model="gpt2")
prompt = "The chicken crossed"
result = generator(prompt, max_new_tokens=40, num_return_sequences=1, truncation=True)
print(result[0]["generated_text"])
```

```python
for temp in [0.2, 0.5, 1.3]:
    result = generator(
        prompt, max_new_tokens=40, do_sample=True, 
        temperature=temp, truncation=True
    )
    print(f"Temperature {temp}:")
    print(result[0]["generated_text"])
    print("---")
```
