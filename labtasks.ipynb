```python
def weather_advice(temperature_c, humidity_percent, wind_speed_kmh, uv_index):
    advice = []

    # Temperature rules
    if temperature_c >= 35:
        advice.append("extreme heat: stay hydrated and avoid the sun.")
    elif temperature_c <= 5 and temperature_c > -10:
        advice.append("cold weather: wear a warm jacket.")

    if temperature_c <= -10:
        advice.append("extreme cold weather: take necessary precautions.")

    if temperature_c >= 50:
        advice.append("extremely hot weather: stay indoors.")

    # Humidity rule
    if humidity_percent >= 80:
        advice.append("high humidity: expect possible rain, carry an umbrella.")

    # Wind speed rule
    if wind_speed_kmh >= 40:
        advice.append("strong wind warning: secure loose objects and avoid unnecessary travel.")

    # UV index rules
    if uv_index >= 11:
        advice.append("extreme UV index: avoid direct sunlight and use strong sun protection.")
    elif uv_index >= 8:
        advice.append("very high UV index: use sunscreen, wear protective clothing, and seek shade.")

    # Default advice
    if not advice:
        advice.append("conditions look moderate. No special precautions needed.")

    return advice


if __name__ == "__main__":
    sample_days = [
      (18, 35, 12, 5),
    (42, 60, 44, 6),
    (-18, 25, 20, 0),
    (31, 85, 38, 11),
    (5, 81, 52, 3)
    ]

    for temp, humidity, wind, uv in sample_days:
        result = weather_advice(temp, humidity, wind, uv)

        print(
            f"Temp={temp}C, Humidity={humidity}%, "
            f"Wind Speed={wind} km/h, UV Index={uv}"
        )
        print(f"Advice: {result}")
        print("-" * 80)
```

```python
# Keyword-counting classification is similar to facial recognition or medical diagnosis as the system looks for patterns/features in input data and uses them to assign a class.

spam_keywords = ["winner","free","urgent","click here","prize","limited time","act now","cash","bonus","offer"]

THRESHOLD = 4


def classify_email(email_text):
    email_lower = email_text.lower()
    count = 0

    for keyword in spam_keywords:
        if keyword in email_lower:
            count += 1

    if count >= THRESHOLD:
        return "Spam", count
    else:
        return "Legitimate", count


# Test emails with their correct labels
test_emails = [
    ("Congratulations! You are a winner! Claim your free prize now.", "Legitimate"),
    ("Urgent! Click here to claim your free prize.", "Spam"),
    ("You are the lucky winner. Click here for your prize.", "Spam"),
    ("Hi, the meeting is scheduled for 3 PM tomorrow.", "Legitimate"),
    ("Thank you for your purchase. Your order has been shipped.", "Legitimate"),
    ("Please review the attached report before Friday.", "Legitimate")
]


correct = 0

for email, correct_label in test_emails:
    predicted_label, keyword_count = classify_email(email)

    if predicted_label == correct_label:
        correct += 1

    print("Email:", email)
    print("Keyword count:", keyword_count)
    print("Predicted:", predicted_label)
    print("Correct label:", correct_label)
    print("-" * 70)


accuracy = (correct / len(test_emails)) * 100

print(f"Correct classifications: {correct}/{len(test_emails)}")
print(f"Classifier accuracy: {accuracy:.2f}%")
```
